# Дефекты металлической поверхности: анализ данных
Сохранённые результаты EDA получены на неизменённых train/val/test. Число изображений и рамок считается отдельно.


```python
from pathlib import Path
import json
root = Path.cwd()
if not (root / 'results/reference/eda').is_dir():
    root = root.parent
eda = json.loads((root / 'results/reference/eda/eda.json').read_text(encoding='utf-8-sig'))
for split, item in eda['dataset'].items():
    print(split, item['images'], 'изображений;', item['objects'], 'рамок')
```

```python
from IPython.display import display, Image
for name in ['class_objects.png', 'box_sizes.png', 'train_examples.png']:
    display(Image(filename=str(root / 'results/reference/eda/figures' / name)))
```
